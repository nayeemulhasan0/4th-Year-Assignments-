```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def vec(expr):
    return sp.Matrix(expr)

def to_tuple(v):
    return tuple(sp.simplify(sp.expand(e)) for e in list(v))

def fs(r, t):
    r1 = sp.diff(r, t); r2 = sp.diff(r1, t); r3 = sp.diff(r2, t)
    n = sp.Matrix.norm; c = sp.Matrix.cross
    s = n(r1)
    kappa = sp.simplify(n(c(r1, r2)) / s**3)
    tau = sp.simplify(sp.Matrix.hstack(r1, r2, r3).det() / n(c(r1, r2))**2)
    T = sp.simplify(r1 / s)
    dTds = sp.simplify(sp.diff(T, t) / s)
    N = sp.simplify(dTds / sp.Matrix.norm(dTds))
    B = sp.simplify(sp.Matrix(c(T, N)))
    return to_tuple(T), to_tuple(N), to_tuple(B), kappa, tau

t = sp.symbols('t', real=True)

r1 = vec([sp.exp(t), sp.exp(t)*sp.cos(t), sp.exp(t)*sp.sin(t)])
T1,N1,B1,k1,ta1 = fs(r1, t)

r2 = vec([2*sp.cos(t), 3*sp.sin(t), 0])
T2,N2,B2,k2,ta2 = fs(r2, t)

print("Curve i:")
print("T =", T1)
print("N =", N1)
print("B =", B1)
print("kappa =", sp.simplify(k1))
print("tau =", sp.simplify(ta1))

print("\nCurve ii:")
print("T =", T2)
print("N =", N2)
print("B =", B2)
print("kappa =", sp.simplify(k2))
print("tau =", sp.simplify(ta2))

k1f = sp.lambdify(t, k1, 'numpy')
k2f = sp.lambdify(t, k2, 'numpy')
tt1 = np.linspace(0, 2*np.pi, 200)
tt2 = np.linspace(0, 2*np.pi, 200)
plt.plot(tt1, k1f(tt1), label='curve i')
plt.plot(tt2, k2f(tt2), label='curve ii')
plt.xlabel('t'); plt.ylabel('kappa'); plt.legend(); plt.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def T(x, y):
    return 3 * x**2 * y

x = np.linspace(-2, 0, 100)
y = np.linspace(0, 2, 100)
X, Y = np.meshgrid(x, y)
Z = T(X, Y)

grad_y, grad_x = np.gradient(Z, y, x)

x0, y0 = -1, 3/2
idx_x = np.argmin(np.abs(x - x0))
idx_y = np.argmin(np.abs(y - y0))
gradient = [round(grad_x[idx_y, idx_x],4), round(grad_y[idx_y, idx_x],4)]

print(f"Gradient at ({x0}, {y0}): {gradient}")

direction = np.array([-1, -1/2])
unit_direction = direction / np.linalg.norm(direction)
D_u = np.dot(gradient, unit_direction)

print(f"Directional derivative: {D_u:.4f}")

fig = plt.figure(figsize=(14, 6))

ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax1.scatter([x0], [y0], [T(x0, y0)], color='red', s=100)
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('T(x,y)')
ax1.set_title('Temperature T(x,y) = 3x²y')

ax2 = fig.add_subplot(122)
Du = grad_x * unit_direction[0] + grad_y * unit_direction[1]
contour = ax2.contourf(X, Y, Du, levels=20, cmap='coolwarm')
ax2.scatter([x0], [y0], color='red', s=100)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('Directional Derivative')
plt.colorbar(contour, ax=ax2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.optimize import fsolve
import matplotlib.pyplot as plt

# Lagrange equations: ∇T = λ∇g
def equations(v):
    x, y, z, L = v
    return [16*x - 8*L*x, 4*z - 2*L*y, 4*y - 16 - 8*L*z, 4*x**2 + y**2 + 4*z**2 - 16]

# Solve with multiple starting points
sols = []
for guess in [[1,2,1,1], [0,4,0,1], [0,-2,-2,1]]:
    x, y, z, L = fsolve(equations, guess)
    if abs(4*x**2 + y**2 + 4*z**2 - 16) < 1e-5:
        T = 8*x**2 + 4*y*z - 16*z + 600
        sols.append((x, y, z, T))

sols.sort(key=lambda s: s[3], reverse=True)
hot = sols[0]

print(f"Hottest point: ({hot[0] }, {hot[1] }, {hot[2] })")
print(f"Max temperature: {hot[3] }°")

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')

u, v = np.linspace(0, 2*np.pi, 30), np.linspace(0, np.pi, 30)
x = 2 * np.outer(np.cos(u), np.sin(v))
y = 4 * np.outer(np.sin(u), np.sin(v))
z = 2 * np.outer(np.ones(len(u)), np.cos(v))

ax.plot_surface(x, y, z, alpha=0.3, color='cyan')
ax.scatter(hot[0], hot[1], hot[2], color='red', s=150)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Hottest Point on Ellipsoid')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.integrate import quad

def integrand(t):
    x = np.cos(t)
    y = np.sin(t)
    z = t
    
    dx_dt = -np.sin(t)
    dy_dt = np.cos(t)
    dz_dt = 1
    
    ds_dt = np.sqrt(dx_dt**2 + dy_dt**2 + dz_dt**2)
    
    function_value = x*y + z**3
    
    return function_value * ds_dt

result = quad(integrand, 0, np.pi)

print(f"The line integral is approximately {result[0]}")
```

```python
import numpy as np
from scipy import integrate

def calculate_cylinder_mass(radius, height):
    
    def integrand(x, y, z):
        return x**2 + y**2
    
    def bounds_z(x, y):
        return [-height/2, height/2]
    
    def bounds_y(x):
        return [-np.sqrt(radius**2 - x**2), np.sqrt(radius**2 - x**2)]
    
    def bounds_x():
        return [-radius, radius]
    
    mass, error = integrate.nquad(
        integrand,
        [bounds_z, bounds_y, bounds_x]
    )
    
    return mass

mass = calculate_cylinder_mass(radius=2, height=3)

print(f"The mass of the cylinder is {mass:.2f} kg")
```

```python
import numpy as np
from scipy import integrate

def calculate_work():
    def integrand(x, y):
        return 3*x**2 + 3*y**2
    
    def bounds(theta):
        x = np.cos(theta)
        y = np.sin(theta)
        return integrand(x, y)
    
    work = integrate.quad(bounds, 0, 2*np.pi)[0]
    
    return work

result = calculate_work()
print(f"The work done by the force field is: {result:.4f}")
```

```python
import numpy as np
from scipy import integrate

def calculate_flux():
    def integrand(z, r, theta):
        x = r * np.cos(theta)
        y = r * np.sin(theta)
        
        divergence = 1 + 3 * (r * np.sin(theta))**2 + 2*z
        
        return divergence * r

    r_lower = 0
    r_upper = 3
    theta_lower = 0
    theta_upper = 2 * np.pi
    z_lower = 0
    z_upper = 2

    result, error = integrate.tplquad(
        integrand,
        theta_lower, theta_upper,
        lambda theta: r_lower, lambda theta: r_upper,
        lambda r, theta: z_lower, lambda r, theta: z_upper
    )
    
    return result

flux = calculate_flux()
print(f"The outward flux is approximately {flux:.4f}")
```

```python
import numpy as np
from scipy import integrate

def line_integral():
    def integrand(t):
        return 8
    return integrate.quad(integrand, 0, 2*np.pi)[0]

def surface_integral():
    def integrand(r, theta):
        return 4 * r
    return integrate.dblquad(integrand, 0, 2*np.pi, 0, 2)[0]

line_result = line_integral()
surface_result = surface_integral()

print(f"Line integral = {line_result:.4f}")
print(f"Surface integral = {surface_result:.4f}")
```
