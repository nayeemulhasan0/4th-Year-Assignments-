## A1Q1



```python
def get_grade(marks):
    if marks >= 80: return "A+", 4.00
    elif marks >= 75: return "A", 3.75
    elif marks >= 70: return "A-", 3.50
    elif marks >= 65: return "B+", 3.25
    elif marks >= 60: return "B", 3.00
    elif marks >= 55: return "B-", 2.75
    elif marks >= 50: return "C+", 2.50
    elif marks >= 45: return "C", 2.25
    elif marks >= 40: return "D", 2.00
    else: return "F", 0.00

roll = float(input("Enter Roll No: "))
first_incourse = float(input("Enter first incourse marks: "))
second_incourse = float(input("Enter second incourse marks: "))
attendance = float(input("Enter attendance marks: "))
final = float(input("Enter final marks: "))

total = ((first_incourse + second_incourse) / 2) + attendance + final

letter_grade, grade_point = get_grade(total)

print("\nRoll | Letter Grade | Grade Point")
print("-" * 35)
print(f"{roll}  | {letter_grade:^11s} | {grade_point:^10.2f}")
```

```python
import re

with open("regex-sum-42.txt", 'r') as file:
    content = file.read()
    numbers = re.findall('\d+', content)

    total_sum = sum(int(i) for i in numbers)

print(f'The sum of all numbers in the file is: {total_sum}')


```

## A1Q3



```python
unique_words = []

with open('amth.txt', 'r') as file:

    for _ in range(1):               ## 
        next(file)

        for line in file:
            words = line.split()
            for word in words:
                if word not in unique_words:
                    unique_words.append(word)

unique_words.sort()

for word in unique_words:
    print(word)


```

```python
# file_name = input("Enter file name: ")
file_name = "mbox-short.txt"

try:
    with open(file_name, 'r') as file:
        values = [float(line.split(":")[1]) for line in file if line.startswith("X-DSPAM-Confidence:")]

    if values:
        print("Average spam confidence:", sum(values) / len(values))
    else:
        print("No matching lines found.")

except FileNotFoundError:
    print("File not found. Please check the filename and try again.")
except ValueError:
    print("Error: File contains improperly formatted lines.")
```

```python
counts = {}
with open('mbox.txt') as file:
    for line in file:
        if line.startswith('From '):
            counts[line.split()[1]] = counts.get(line.split()[1], 0) + 1
sender = max(counts, key=counts.get)
print(sender, counts[sender])

```
